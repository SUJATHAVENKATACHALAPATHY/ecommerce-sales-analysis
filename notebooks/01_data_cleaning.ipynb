```python
import pandas as pd
import numpy as np

path = "../data/raw/"

orders = pd.read_csv(path + "olist_orders_dataset.csv")
customers = pd.read_csv(path + "olist_customers_dataset.csv")
items = pd.read_csv(path + "olist_order_items_dataset.csv")
payments = pd.read_csv(path + "olist_order_payments_dataset.csv")
reviews = pd.read_csv(path + "olist_order_reviews_dataset.csv")
products = pd.read_csv(path + "olist_products_dataset.csv")
sellers = pd.read_csv(path + "olist_sellers_dataset.csv")
categories = pd.read_csv(path + "product_category_name_translation.csv")

orders.head()
```

```python
for name, df in {"orders": orders, "customers": customers, "items": items,
                 "payments": payments, "reviews": reviews,
                 "products": products, "sellers": sellers}.items():
    print(f"{name}: {df.shape[0]:,} rows, {df.shape[1]} columns")

orders.info()
```

```python
for name, df in {"orders": orders, "customers": customers, "items": items,
                 "payments": payments, "reviews": reviews,
                 "products": products, "sellers": sellers}.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

```python
date_cols = ["order_purchase_timestamp", "order_approved_at",
             "order_delivered_carrier_date", "order_delivered_customer_date",
             "order_estimated_delivery_date"]

for col in date_cols:
    orders[col] = pd.to_datetime(orders[col])

orders["purchase_year"] = orders["order_purchase_timestamp"].dt.year
orders["purchase_month"] = orders["order_purchase_timestamp"].dt.month
orders["purchase_year_month"] = orders["order_purchase_timestamp"].dt.to_period("M").astype(str)

orders["delivery_days"] = (orders["order_delivered_customer_date"]
                           - orders["order_purchase_timestamp"]).dt.days
orders["delay_days"] = (orders["order_delivered_customer_date"]
                        - orders["order_estimated_delivery_date"]).dt.days

orders[["order_purchase_timestamp", "purchase_year_month", "delivery_days", "delay_days"]].head()
```

```python
orders.dtypes
```

```python
products["product_category_name"] = products["product_category_name"].fillna("unknown")

products = products.merge(categories, on="product_category_name", how="left")
products["product_category_name_english"] = products["product_category_name_english"].fillna("unknown")

products[["product_id", "product_category_name", "product_category_name_english"]].head()
```

```python
reviews["review_creation_date"] = pd.to_datetime(reviews["review_creation_date"])
reviews = (reviews.sort_values("review_creation_date")
                  .drop_duplicates("order_id", keep="last"))
reviews.shape
```

```python
# total payment per order
order_payments = payments.groupby("order_id", as_index=False)["payment_value"].sum()

df = (items
      .merge(orders, on="order_id", how="left")
      .merge(customers, on="customer_id", how="left")
      .merge(products[["product_id", "product_category_name_english"]], on="product_id", how="left")
      .merge(sellers, on="seller_id", how="left")
      .merge(reviews[["order_id", "review_score"]], on="order_id", how="left"))

df["revenue"] = df["price"] + df["freight_value"]
df.shape
```

```python
df.to_csv("../data/cleaned/ecommerce_cleaned.csv", index=False)
orders.to_csv("../data/cleaned/orders_cleaned.csv", index=False)
print("Saved")
```
