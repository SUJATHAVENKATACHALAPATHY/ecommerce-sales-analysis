```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/cleaned/olist.db")

tables = {"orders": "olist_orders_dataset", "customers": "olist_customers_dataset",
          "items": "olist_order_items_dataset", "payments": "olist_order_payments_dataset",
          "reviews": "olist_order_reviews_dataset", "products": "olist_products_dataset",
          "sellers": "olist_sellers_dataset", "categories": "product_category_name_translation"}

for table, file in tables.items():
    pd.read_csv(f"../data/raw/{file}.csv").to_sql(table, conn, if_exists="replace", index=False)

pd.read_sql("SELECT name FROM sqlite_master WHERE type='table'", conn)
```
